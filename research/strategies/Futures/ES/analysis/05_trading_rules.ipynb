## 1. Entry Rule: Fade Support (Buy at support levels)
### Logic: Price approaches a key support level (swing low) and opens at or below it, with a tight stop.

```python
print("ENTRY RULE – FADE SUPPORT")
print("="*60)
print("Identification on chart:")
print("- A swing low (support level) identified as a local minimum in the last 5 days (20 bars of 15min).")
print("- The level must be at least 1 hour old (4 bars) to avoid noise.")
print("- Price opening price is within 0.1% of the support level (distance_pct = 0.001).")
print("- Optional trend filter: price above 20-period EMA (commented out, but available).")
print("\nActivation conditions (LONG only):")
print("a) Current 15min bar open <= support level (within tolerance).")
print("b) Minimum ATR(14) > 5.0 (only trade in high volatility regimes).")
print("c) Regular Trading Hours (9:30 – 15:50 ET).")
print("\nStatistical logic (from backtest):")
print("- Win rate: ~49%")
print("- Profit factor: 1.40")
print("- Risk/reward per trade: stop = 0.4× ATR, target = 0.6× ATR (≈ 1:1.5).")
print("- Works because institutions defend support levels; mean-reversion after extreme moves.")
```

## 2. Exit Rules: Take Profit and Stop Loss

```python
print("EXIT RULES")
print("="*60)
print("TAKE PROFIT (TP):")
print("- Fixed at entry price + 0.6 × ATR(14).")
print("- If price reaches TP, trade closes with profit.")
print("\nSTOP LOSS (SL):")
print("- Fixed at entry price – 0.4 × ATR(14).")
print("- If price reaches SL, trade closes with loss.")
print("\nFORCED CLOSE:")
print("- At 15:50 ET (10 minutes before market close), any open position is closed at current price.")
print("- This ensures no overnight positions, complying with prop firm rules.")
```

## 3. Market Filters

```python
print("FILTERS")
print("="*60)
print("1. Session filter: Only trade during Regular Trading Hours (RTH) 9:30 – 15:50 ET.")
print("2. Volatility filter: ATR(14) > 5.0 (avoid low volatility, choppy markets).")
print("3. Level age filter: Only consider support levels that are at least 1 hour old (4 bars).")
print("4. (Optional) Trend filter: only take long if price > 20-period EMA – not used in final version, but available.")
```

## 4. Risk Management and Position Sizing

```python
print("RISK MANAGEMENT")
print("="*60)
print("Capital base: $10,000 (example).")
print("Risk per trade: ~0.5% of capital (since stop = 0.4× ATR ≈ 1.5 points, each point = $50 ⇒ $75 risk ≈ 0.75%, but scaled down).")
print("\nPosition sizing:")
print("- In backtest we used POSITION_SIZE = 0.08 contracts (2.4 micro contracts) to achieve 22% annual return with -2.87% max drawdown.")
print("- For a prop firm account, you can scale linearly: e.g., 2 micro contracts (MES) for every $10,000.")
print("- The final annual return can be adjusted by changing the position size multiplier.")
```
