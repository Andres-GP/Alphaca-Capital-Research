## 📖 Mi historia de trading

Empecé buscando una estrategia simple, robusta y que pudiera operar sin estrés. Después de analizar años de datos del ES, descubrí que **los niveles de soporte son defendidos por grandes instituciones**, y que cuando el precio toca un soporte importante después de un movimiento brusco, suele rebotar.

Mi estrategia se llama **"El Cazador de Soporte"**: espero a que el precio llegue a zonas donde históricamente ha rebotado, y entro con una stop ajustado y un objetivo pequeño pero consistente.

No opero en rangos laterales ni cuando la volatilidad es baja. Solo actúo cuando el mercado me da una señal clara: precio en un soporte relevante, con volatilidad suficiente, y dentro del horario regular.

Segun el backtesting esta da un 22% anualizado durante los últimos 6 años, con un drawdown máximo de solo -2.87%. Lo más importante: **nunca dejo posiciones abiertas overnight**, lo que me permite dormir tranquilo y cumplir con las reglas de la fondeadora.

## 🎯 Metáfora: "El pescador de truchas"

Imagina que estás en un río y ves una trucha grande que siempre sube a la superficie cerca de una misma roca. No persigues la trucha, esperas junto a la roca. Cuando la trucha aparece, lanzas el anzuelo. Si no pica en unos minutos, recoges el anzuelo y esperas otra oportunidad.

Mi estrategia es así: identifico “rocas” (niveles de soporte) donde el precio ha rebotado en el pasado. Espero pacientemente a que el precio llegue a esa roca. Entonces lanzo mi orden de compra. Pongo un stop muy ajustado (como un sedal corto) para que si no es la trucha correcta, pierda solo la carnada, no el equipo. Pongo un objetivo pequeño pero seguro: no quiero la trucha más grande, solo una captura constante.

Si en media hora no ha picado, recojo el anzuelo (time stop). Y lo más importante: **al atardecer (15:50) guardo todo y me voy a casa**, no dejo anzuelos en el agua fuera de hora.

## ✅ Checklist operativo para tiempo real

Antes de abrir una operación, reviso mentalmente esta lista:

- [ ] **¿Es horario RTH?** (9:30 – 15:50 ET) → si no, no opero.
- [ ] **¿Hay un soporte válido?** (mínimo local de al menos 20 velas de 15m de antigüedad y al menos 1 hora de antigüedad)
- [ ] **¿El precio de apertura de la vela actual está a menos de 0.1% del soporte?**
- [ ] **¿ATR(14) > 5?** (suficiente volatilidad)
- [ ] **(Opcional) ¿El precio está por encima de la EMA de 20 periodos?** (filtro de tendencia)
- [ ] **¿No hay noticias importantes en los próximos 15 minutos?** (evitar FOMC, CPI, NFP)

Si todo responde SÍ: **ENTRO LARGO** al precio de apertura.
Stop: entrada – 0.4× ATR.
Take profit: entrada + 0.6× ATR.

**Gestión de la operación:**
- Si no se alcanza TP ni SL en 2 horas, cierro manualmente (time stop).
- A las 15:50 ET, cierro cualquier posición que aún esté abierta (al precio de mercado).

**Después de cerrar:** anoto en mi diario: resultado, observaciones, emociones.

## 💰 Reglas de gestión de riesgo y capital

**Riesgo por operación:** máximo 0.5% del capital total (ej. con $10,000, el stop loss no debe exceder $50).  
Como cada punto en ES vale $50, un stop de 1 punto equivale a $50 = 0.5% de riesgo.  
Nuestro stop típico es 0.4× ATR ≈ 2-3 puntos → riesgo ≈ 1-1.5% (dependiendo del ATR).  
Para ajustarlo al 0.5%, reducimos el tamaño de la posición:

- **Para capital de $10,000:** usamos 0.2 contratos (2 micros MES) → cada punto = $10 → el riesgo de 3 puntos = $30 = 0.3% (muy seguro).
- **Escalamiento:** por cada $5,000 adicionales, añadir 1 micro contrato.

**Objetivo de retiro de ganancias:** mensual, retirar el 50% de las ganancias netas; el resto se reinvierte (aumentando tamaño gradualmente).

**Límites diarios:** si pierdo 1.5% del capital en un día, detengo la operativa y lo reviso al día siguiente.  
**Límite semanal:** si el drawdown semanal supera el 3%, reduzco el tamaño a la mitad la semana siguiente.

## 🔁 Adaptación al cambio (elemento dinámico)

**Escenario simulado: Un cambio significativo en la volatilidad (ej. VIX sube de 15 a 30 en una semana).**

Mi estrategia original solo entra con ATR > 5. En alta volatilidad, el ATR puede subir a 10 o más. Los stops basados en 0.4× ATR serían muy amplios (4-5 puntos), y el riesgo por operación se dispararía.

**Adaptación:**
1. Reducir el multiplicador de stop a 0.25× ATR (stop más ajustado) para mantener el riesgo similar.
2. Reducir el tamaño de la posición a la mitad.
3. Aumentar la distancia de entrada al soporte a 0.2% (para no pillar falsos).

Validaría la adaptación haciendo un backtest específico sobre periodos de alta volatilidad (2022, 2020) con los nuevos parámetros, y haría walk‑forward sobre los últimos 6 meses. Si el profit factor sigue ≥1.3 y drawdown <8%, implementaría el cambio.
